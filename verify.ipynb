```python
import matplotlib.pyplot as plt
from verify import *
from pprint import pprint
```

```python
def verify_and_display(img1_path, img2_path, checkpoint):
    model = load_model(checkpoint)
    result = verify(img1_path, img2_path, model)
    pprint(result)

    fig, axes = plt.subplots(2, 2, figsize=(12, 10))

    img1 = plt.imread(img1_path)
    img2 = plt.imread(img2_path)

    axes[0, 0].imshow(img1)
    axes[0, 0].axis('off')
    axes[0, 0].set_title(f'Image 1\nAge: {result["age1"]} years', fontsize=14, fontweight='bold')

    axes[0, 1].imshow(img2)
    axes[0, 1].axis('off')
    axes[0, 1].set_title(f'Image 2\nAge: {result["age2"]} years', fontsize=14, fontweight='bold')

    axes[1, 0].axis('off')
    result_text = "✓ SAME PERSON" if result["verified"] else "✗ DIFFERENT PERSONS"
    result_color = "green" if result["verified"] else "red"
    axes[1, 0].text(0.5, 0.5, result_text, 
                    ha='center', va='center', 
                    fontsize=24, fontweight='bold', 
                    color=result_color,
                    transform=axes[1, 0].transAxes)

    axes[1, 1].axis('off')
    metrics_text = f"Distance: {result['distance']:.4f}\nThreshold: {result['threshold']}"
    axes[1, 1].text(0.5, 0.5, metrics_text,
                    ha='center', va='center',
                    fontsize=24, fontweight='bold',
                    color='black',
                    transform=axes[1, 1].transAxes)

    plt.tight_layout()
    plt.show()
    return result
```

```python
img1_path = "dataset/image1.png"
img2_path = "dataset/image2.png"
checkpoint = "best.pth"

result = verify_and_display(img1_path, img2_path, checkpoint)
```
